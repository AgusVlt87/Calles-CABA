```python
import pandas as pd

# leemos la columna "agrupacion_nombre" de todos los datasets electorales que tenemos dentro de data/elecciones para ver el unique de agrupaciones que hay
def obtener_agrupaciones_unicas():
    # obtenemos la lista de archivos dentro de data/elecciones
    import os
    archivos = os.listdir("data/elecciones")
    
    # leemos cada archivo y obtenemos el unique de agrupaciones
    agrupaciones_unicas = set()
    for archivo in archivos:
        df = pd.read_csv(f"data/elecciones/{archivo}")
        print(f"Archivo: {archivo}, Agrupaciones únicas: {df['agrupacion_nombre'].unique()}")
        print(f'--' * 20)
    
    return agrupaciones_unicas

obtener_agrupaciones_unicas()
```
